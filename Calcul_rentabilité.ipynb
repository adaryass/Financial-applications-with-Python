```python
import numpy as np
import pandas as pd

# Create a synthetic price series for an asset (e.g., daily close)
np.random.seed(42)
days = 30

# Dates
dates = pd.date_range(end=pd.Timestamp.today(), periods=days)

mu = 0.0008
sigma = 0.01

# Simulated returns (here treated as log-returns)
returns = np.random.normal(loc=mu, scale=sigma, size=days)

price_0 = 100.0

# Price path from log-returns: P_t = P0 * exp(cumsum(r_t))
prices = price_0 * np.exp(np.cumsum(returns))

# DataFrame
df = pd.DataFrame({'price': prices}, index=dates)
df.index.name = 'date'
df.head()
```

```python
prices = []

price = price_0
for r in returns:
    price *= (1 + r)
    prices.append(price)
```

```python
df2 = pd.DataFrame({'price': prices}, index = dates)
df2.index.name = 'date'

df2.head(), df2.tail()
```

```python
import yfinance as yf
import numpy as np
import pandas as pd

# Télécharger les données Close de Google
df = yf.download("GOOGL", start="2020-01-01", interval="1d")

price_series = df['Close'].dropna()

# Conversion en array NumPy
prices_arr = price_series.to_numpy()
prices_arr
```

```python
# Rentabilité simple
returns = prices_arr[1:] / prices_arr[:-1] - 1

# Rentabilité logarithmique
log_returns = np.log(prices_arr[1:] / prices_arr[:-1])

# Affichage
returns[:5], log_returns[:5]
```
